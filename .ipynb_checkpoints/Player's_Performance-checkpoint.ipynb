## QAULITY REPORT

```python
def quality_report(df):
    """
    Description: Displays quality of data in terms of missing values, 
    unique numbers, datatypes etc.
    
    Arguments: Dataframe
    """
    dtypes = df.dtypes
    nuniq = df.T.apply(lambda x: x.nunique(), axis=1)
    total = df.isnull().sum().sort_values(ascending = False)
    percent = (df.isnull().sum()/df.isnull().count()*100).sort_values(ascending = False)
    quality_df  = pd.concat([total, percent, nuniq, dtypes], axis=1, keys=['Total NaN', 'Percent of NaN','Nunique', 'Dtype'])
    display(quality_df)
quality_report(df)
```

### Questions I want to ask having know what my data looks like


### **Key Questions.**

    1) Are pitchers reliably creating strikes via command or stuff? (CSW%)  
    2) Do we miss bats when hitters swing? (Whiff/Swing%)  
    3) Are we living in the zone without leaking walks? (Balls%)  
    4) Is pitch mix balanced and purposeful? (Pitch-type share)  
    5) Which batters actually produced in these games? (OPS, H/BB/K)  
    6) What does each hitter’s approach look like? (Swing%, Contact%, Whiff/Swing%)

## **Methods.**


    - Normalize `PitchCall` into swings, whiffs, balls, called strikes, balls in play, hits, walks, strikeouts; define terminal events ≈ 'PA'.
    
    - Aggregate by 'PitcherId' and 'BatterId' for transparent, coach-facing rates (CSW%, Whiff/Swing%, Swing%, Contact%) and **OPS** (OBP + SLG) approximations.
    
    - Build 'pitch mix' shares by pitcher; minimum sample filters to limit noise.
    
    - Visualize with  bar charts (CSW leaders, OPS leaders) and stacked mix.


## Define the Events Set

```python

HIT_CALLS = {"single","double","triple","home_run"}
OUT_INPLAY_CALLS = {
    "field_out","groundout","flyout","lineout","pop_out","force_out",
    "double_play","grounded_into_double_play","sac_fly","sac_bunt","other_out"
}
SWINGING = {"swinging_strike"}
CALLED_STRIKE = {"called_strike"}
BALLS = {"ball","blocked_ball","pitchout"}
WALKS = {"walk","intent_walk"}
STRIKEOUTS = {"strikeout","strikeout_double_play"}
```

## Define the Flags

```python

df["is_hit"] = df["PitchCall"].isin(HIT_CALLS).astype(int)
df["is_out_bip"] = df["PitchCall"].isin(OUT_INPLAY_CALLS).astype(int)
df["is_bip"] = ((df["is_hit"]==1) | (df["is_out_bip"]==1)).astype(int)
df["is_swing"] = (df["PitchCall"].isin(SWINGING.union({"foul"})) | (df["is_bip"]==1)).astype(int)
df["is_whiff"] = df["PitchCall"].isin(SWINGING).astype(int)
df["is_called_strike"] = df["PitchCall"].isin(CALLED_STRIKE).astype(int)
df["is_ball"] = df["PitchCall"].isin(BALLS).astype(int)
df["is_walk"] = df["PitchCall"].isin(WALKS).astype(int)
df["is_strikeout"] = df["PitchCall"].isin(STRIKEOUTS).astype(int)
df["is_terminal"] = df["PitchCall"].isin(HIT_CALLS | OUT_INPLAY_CALLS | WALKS | STRIKEOUTS).astype(int)
```

## Batter Summary

```python

def batter_summary(df):
    g = df.groupby("BatterId", dropna=False)
    out = g.agg(
        pitches=("PitchId","count"),
        swings=("is_swing","sum"),
        whiffs=("is_whiff","sum"),
        called_strikes=("is_called_strike","sum"),
        balls=("is_ball","sum"),
        bip=("is_bip","sum"),
        hits=("is_hit","sum"),
        walks=("is_walk","sum"),
        strikeouts=("is_strikeout","sum"),
        pa=("is_terminal","sum"),
        singles=("hit_single","sum"),
        doubles=("hit_double","sum"),
        triples=("hit_triple","sum"),
        homers=("hit_home_run","sum"),
    ).reset_index()
    out["AB"] = (out["pa"] - out["walks"]).clip(lower=0)
    out["contacts"] = (out["swings"] - out["whiffs"]).clip(lower=0)
    out["H"] = out["singles"] + out["doubles"] + out["triples"] + out["homers"]
    out["TB"] = (out["singles"]*1 + out["doubles"]*2 + out["triples"]*3 + out["homers"]*4)
    out["OBP"] = (out["H"] + out["walks"]) / out["pa"].replace(0, np.nan)
    out["SLG"] = out["TB"] / out["AB"].replace(0, np.nan)
    out["OPS"] = out["OBP"] + out["SLG"]
    out["Swing%"] = out["swings"] / out["pitches"]
    out["Contact%"] = out["contacts"] / out["swings"].replace(0, np.nan)
    out["Whiff/Swing%"] = out["whiffs"] / out["swings"].replace(0, np.nan)
    return out
```

```python
batter_summary(df)
```

## Pitcher Summary

```python
def pitcher_summary(df):
    g = df.groupby("PitcherId", dropna=False)
    out = g.agg(
        pitches=("PitchId","count"),
        swings=("is_swing","sum"),
        whiffs=("is_whiff","sum"),
        called_strikes=("is_called_strike","sum"),
        balls=("is_ball","sum"),
        bip=("is_bip","sum"),
        hits=("is_hit","sum"),
        walks=("is_walk","sum"),
        strikeouts=("is_strikeout","sum"),
        pa=("is_terminal","sum"),
    ).reset_index()
    out["CSW%"] = (out["called_strikes"] + out["whiffs"]) / out["pitches"]
    out["Whiff/Swing%"] = out["whiffs"] / out["swings"].replace(0, np.nan)
    out["Balls%"] = out["balls"] / out["pitches"]
    out["BIP%"] = out["bip"] / out["pitches"]
    return out
```

```python
pitcher_summary(df)
```

## Pitch Mix

```python

mix = (
    df.groupby(["PitcherId","PitchType"], dropna=False)["PitchId"]
      .count().rename("count").reset_index()
)
mix["share"] = mix.groupby("PitcherId")["count"].transform(lambda s: s / s.sum())
```

```python
mix
```

# Checking the Pitch-type usage & quality.

I compute each pitcher’s mix (counts and % share by pitch type) and join it with CSW by pitch type to assess effectiveness. 

I then try to derive an opportunity score = CSW × (1 − share) to show underused but effective pitches. This is actionable in baseball because if a slider shows high CSW at low usage, one can increase its share in relevant counts/splits. Conversely, high-usage/low-CSW shapes are candidates to trim or re-shape.

```python
# CSW% by pitch type (per pitcher)
df["is_csw"] = df["PitchCall"].isin({"called_strike","swinging_strike"}).astype(int)

csw_by_type = (
    df.groupby(["PitcherId","PitchType"])
      .agg(pitches=("PitchId","count"), csw=("is_csw","mean"))
      .reset_index()
)

mix_q = mix.merge(csw_by_type, on=["PitcherId","PitchType"], how="left")

mix_q["opportunity"] = mix_q["csw"] * (1 - mix_q["share"])
```

```python
mix_q
```

```python
MIN_PITCHTYPE_COUNT = 10 

mix_q_filt = (
    mix_q[mix_q["count"] >= MIN_PITCHTYPE_COUNT]
      .dropna(subset=["PitchType", "csw"])    
      .copy()
)
title_suffix = f"(min {MIN_PITCHTYPE_COUNT} pitches per type)"
```

# Top opportunities bar Charts

```python


topk = (
    mix_q_filt.sort_values(["PitcherId","opportunity"], ascending=[True, False])
              .groupby("PitcherId").head(2)    # top 1–2 levers per pitcher
)

if not topk.empty:
    topk = topk.sort_values("opportunity", ascending=True)
    topk["label"] = topk.apply(
        lambda r: f"P{int(r['PitcherId'])}-{str(r['PitchType'])}", axis=1
    )

    plt.figure()
    plt.barh(topk["label"], topk["opportunity"])
    plt.xlabel("Opportunity = CSW × (1 − Share)")
    plt.title(f"Top Usage Opportunities by Pitcher & Pitch Type {title_suffix}")
    plt.tight_layout()
    # plt.savefig("top_opportunities_barh.png", dpi=150)
    plt.show()
```

### This Bars shows the opportunity which = CSW × (1 − share). High bar means pitch is effective (high CSW) but under-used.

    P11–CH is the clearest lever (highest bar): the changeup earns strikes (called + whiff) and isn’t used enough.

    P1–SL (and P1–CU) both pop: his breakers are playing; usage can be bumped situationally.

    P6–KC (knuckle/curve) shows strong upside if he can land it early/late.

    P7–CU/FC, P5–SL, P9–SL: secondary shapes with headroom.

    P14–FF, P15–FF are at the bottom: either already used enough for their quality or simply not good CSW, so he shouldn’t force more.

## CSW heatmap

```python


pivot = mix_q_filt.pivot_table(index="PitcherId",
                               columns="PitchType",
                               values="csw",
                               aggfunc="mean").sort_index()

if pivot.shape[0] >= 2 and pivot.shape[1] >= 2:
    fig = plt.figure()
    ax = plt.gca()
    im = ax.imshow(pivot.values, aspect="auto")
    ax.set_xticks(range(pivot.shape[1]))
    ax.set_xticklabels([str(c) for c in pivot.columns], rotation=0)
    ax.set_yticks(range(pivot.shape[0]))
    ax.set_yticklabels([str(i) for i in pivot.index])
    ax.set_title(f"CSW by Pitcher × Pitch Type {title_suffix}")
    plt.colorbar(im, ax=ax, label="CSW")
    plt.tight_layout()
    # plt.savefig("csw_heatmap_pitcher_by_type.png", dpi=150)
    plt.show()
```

## CSW heatmap (min 10/ptype)

    Cells are CSW for (Pitcher, PitchType). Brighter = more called + swinging strikes.

    P6–KC is the brightest cell on the grid (excellent strike creation).

    P11–CH and P11–FF both strong: CH is the feature; FF supports it.

    P9–SL and P1–SL: Both are solid legit two-strike options.

    Several SI cells are darker → sinkers here aren’t winning many called/swinging strikes (may be contact-oriented; I will definitely check BIP damage before conclusions).

    P15–FF is very dark → not creating strikes; reassess role/usage.

## Actionable recommendations

These are game-ready levers; Only treated as descriptive for two games, and confirm with count/platoon splits and on-contact damage before big usage changes.

    Increase CH for P11 (primary lever) by +5–10 share pts ahead/2-strike and vs opposite-hand hitters; pair FF up/edge to steal calls because CH has top-tier CSW with room to scale (high opportunity).

    Increase SL for P1 (put-away lever) by +5–8 share pts in two-strike/down-edge lanes; use CU as show-me early/below zone after FF because SL shows strong CSW and both breakers are under-utilized.

    Increase KC for P6 (primary lever) by +5–8 share pts early to steal strikes and late when ahead because KC has elite CSW with low current share (high opportunity).

    Increase CU/FC for P7 (count levers) by +4–6 share pts in neutral/plus counts; keep SL for chase only because CU/FC show better CSW and are under-used.

    Hold or reduce FF for P14 & P15; prioritize breakers or adjust ride/shape/location before scaling FF because FF CSW is poor with limited upside.
    
    
In summary: Scale up under-used winners—pitches with high CSW at low usage, while trimming low-CSW four-seams. With a min 10 pitches/type filter, I recommend: P11–CH +5–10pts, P1–SL +5–8pts (CU early), P6–KC +5–8pts, P7–CU/FC +4–6pts (SL chase), P14 & P15 FF hold/decrease.

### Pitchers CSW components (Called vs Swinging) stacked bars

```python
MIN_P = 15
need = ["PitcherId","pitches","called_strikes","whiffs"]
have_cols = [c for c in need if c in pitchers.columns]
assert len(have_cols) == len(need), "pitchers table must include called_strikes and whiffs."

csw_comp = pitchers.loc[pitchers["pitches"]>=MIN_P, need].copy()
csw_comp["called_rate"]  = csw_comp["called_strikes"] / csw_comp["pitches"]
csw_comp["swinging_rate"]= csw_comp["whiffs"] / csw_comp["pitches"]
csw_comp["csw"] = csw_comp["called_rate"] + csw_comp["swinging_rate"]

# top by CSW
csw_comp = csw_comp.sort_values("csw", ascending=False).head(10).reset_index(drop=True)

x = np.arange(len(csw_comp))
plt.figure()
plt.bar(x, csw_comp["called_rate"], label="Called Strike share")
plt.bar(x, csw_comp["swinging_rate"], bottom=csw_comp["called_rate"], label="Swinging Strike share")
plt.xticks(x, csw_comp["PitcherId"])
plt.ylabel("Share of Total Pitches")
plt.title(f"CSW Components by Pitcher (min {MIN_P} pitches)")
plt.legend()
plt.tight_layout()
plt.show()
```

### Hitters OPS bar with Contact% overlay (and PA labels)

```python
MIN_PA = 5
topN = 12

cols = ["BatterId","pa","OBP","SLG","OPS","Swing%","Contact%","Whiff/Swing%"]
board = (batters.loc[batters["pa"]>=MIN_PA, cols]
                 .sort_values("OPS", ascending=False)
                 .head(topN)
                 .reset_index(drop=True))

x = np.arange(len(board))
plt.figure()
plt.bar(x, board["OPS"])
plt.plot(x, board["Contact%"], marker="o")  # overlay line+markers for Contact%

plt.xticks(x, board["BatterId"])
plt.ylabel("OPS (bars) & Contact% (line)")
plt.title(f"Top Batters by OPS (min {MIN_PA} PA) — with Contact% overlay")

# PA labels on bars
for i, (ops, pa) in enumerate(zip(board["OPS"], board["pa"])):
    plt.text(i, ops + 0.02, f"PA {int(pa)}", ha="center", va="bottom", fontsize=8)

plt.tight_layout()
plt.show()
```
